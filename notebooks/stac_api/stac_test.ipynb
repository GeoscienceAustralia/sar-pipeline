```python
from pystac_client import Client
from odc.stac import load, configure_s3_access

# configure boto3 to use the right profile by default if needed
# configure_s3_access("dev2")
```

```python
# open the STAC API client
client = Client.open("http://0.0.0.0:8082/")
```

```python
# get the first collection from the API
collection = client.search(collections=["s1-nrb-iw-collection"], limit=1)
```

```python
collection.get_all_items()
```

```python
# Query for a specific item by ID

item = client.search(
    collections=["s1-nrb-iw-collection"],
    ids=["ga_s1a_nrb_0-1-0_T055-117597-IW3_20150525T120605Z"],
).get_all_items()[0]
```

```python
item
```

```python
# Load the item as an xarray dataset using odc.stac

ds = load([item], resolution=50)
```

```python
ds
```

```python
# Plot the dataset using xarray's built-in plotting capabilities

ds.hh_gamma0.plot.imshow(col="time", col_wrap=2)
```
